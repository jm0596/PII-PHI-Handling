# PII Suppressed Customers View

Creates a suppression view over the customers table that exposes only the non-PII columns (customer_id, country), then previews the result.

```sql
-- DBTITLE 1,Created View
create or replace view pii_demo.pii_target.customers_suppressed
as select 
  customer_id,
  country
from pii_demo.pii_target.customers
```

```sql
select * from pii_demo.pii_target.customers_suppressed
```
