# RBAC Row Filters

Apply a row filter to the customers table so users only see rows for their permitted country, while members of the `manager` group see all rows.

```sql
-- View all rows before applying the row filter
SELECT * FROM pii_demo.pii_target.customers
```

```sql
-- Row filter function: managers see all countries, everyone else only India
CREATE OR REPLACE FUNCTION pii_demo.access_control.country_filter(country STRING)
RETURN
  CASE
    WHEN is_account_group_member('manager') THEN TRUE
    WHEN country = 'India' THEN TRUE
    ELSE FALSE
  END;
```

```sql
-- Attach the row filter to the customers table on the country column
ALTER TABLE pii_demo.pii_target.customers
SET ROW FILTER pii_demo.access_control.country_filter ON (country);
```

```sql
-- View rows after the filter: managers see all, others see only India
SELECT * FROM pii_demo.pii_target.customers
```
