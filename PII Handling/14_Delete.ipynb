# Delete PII Records

Permanently remove an employee record from the `employees_rounded` table, verify the deletion, and use Delta time travel and VACUUM to manage how long deleted data remains recoverable.

```sql
-- Extend deleted file retention before deleting (default is 7 days)
ALTER TABLE pii_demo.pii_target.employees_rounded SET TBLPROPERTIES ('delta.deletedFileRetentionDuration' = '30 days');
```

```sql
-- Delete the employee record with employee_id = 2
delete from pii_demo.pii_target.employees_rounded where employee_id=2
```

```sql
-- Verify the record is no longer visible in the current table version
select * from pii_demo.pii_target.employees_rounded where employee_id=2
```

```sql
-- Inspect the table's change history to see the delete operation
describe history pii_demo.pii_target.employees_rounded
```

```sql
-- Time travel: the deleted record is still readable in version 1 until files are vacuumed
select * from pii_demo.pii_target.employees_rounded version as of 1 where employee_id=2
```

```sql
-- Remove data files no longer referenced, making the deleted record unrecoverable
vacuum pii_demo.pii_target.employees_rounded
```
